# **Project Summary -**

This project presents a comprehensive analysis of an Amazon Prime TV Shows and Movies dataset using Python-based data analytics techniques. The primary objective of the project is to understand the composition, quality, popularity, and performance of content available on the platform and to convert these findings into meaningful business insights. The analysis was performed using Python, Pandas, NumPy, Matplotlib, and Seaborn, covering data cleaning, data wrangling, exploratory data analysis (EDA), statistical analysis, and data visualization.

The project began with an extensive data cleaning and preprocessing process. The dataset contained approximately 9,868 titles and several important attributes, including content type, release year, genre, age certification, IMDb score, IMDb votes, TMDB score, and TMDB popularity. Missing values, duplicate records, empty genre lists, and highly skewed numerical variables were investigated. The seasons column contained more than 86% missing values, so it was removed because it provided limited analytical value. IMDb scores and votes were analyzed for missing values and distribution patterns, while median imputation was applied to highly skewed variables such as IMDb votes and TMDB popularity. Genre data was also transformed and exploded to enable detailed genre-level analysis.

The content distribution analysis revealed that the platform is strongly dominated by Movies, which account for approximately 86.2% of the dataset, while TV Shows account for only 13.8%. This indicates a significant opportunity to expand the TV Show segment and create a more balanced content portfolio. Genre analysis showed that Drama, Comedy, and Thriller are among the most represented genres. At the same time, genres such as Documentary, History, Reality, Sport, and Music showed potential as underrepresented or underperforming categories that could be explored for targeted content acquisition and promotion.

The project also examined content quality using IMDb and TMDB ratings. Most titles were concentrated around moderate rating ranges, generally between 5 and 7, while highly rated content represented a smaller portion of the overall library. The correlation analysis provided an important insight: IMDb Score and TMDB Score have a moderate positive correlation of approximately 0.49, suggesting that highly rated titles on IMDb tend to receive relatively higher ratings on TMDB as well. However, the relationship between ratings and popularity was considerably weaker. This demonstrates that content quality and popularity are not identical measures.

The analysis of IMDb votes and TMDB popularity further supports this finding. Titles with a large number of votes are not necessarily the highest-rated titles, while highly popular titles do not always have the highest scores. Therefore, relying on only one performance metric could lead to inaccurate business decisions. A combination of ratings, votes, popularity, genre, content type, and audience characteristics provides a more reliable basis for evaluating content performance.

Another important part of the project focused on actor and director performance. High-performing directors and actor-director collaborations were identified using average IMDb scores. This analysis can help identify creative partnerships that have historically produced well-rated content. However, collaboration count should also be considered because a high average score based on only one title may not represent consistent performance.

The age-certification analysis provided additional insights into audience segmentation. Genres such as Horror, Crime, and Thriller showed a strong presence of mature-rated content, whereas Animation and Family content were more associated with family-friendly certifications. These findings can support targeted marketing, personalized recommendations, and audience-specific content strategies.

Overall, the analysis demonstrates that Amazon Prime can benefit from a data-driven, multi-metric content strategy. The platform should continue investing in successful movie content while gradually expanding high-quality TV Shows. It should also identify highly rated but less-popular titles for targeted promotion, study successful actor-director collaborations, and use genre and certification patterns for better audience targeting.

The project concludes that combining IMDb Score, IMDb Votes, TMDB Score, TMDB Popularity, Genre, Age Certification, Content Type, and Actor-Director Performance can provide a stronger foundation for content acquisition, recommendation, marketing, and audience engagement decisions. These insights can ultimately contribute to better content quality, improved user satisfaction, stronger engagement, and higher customer retention..
